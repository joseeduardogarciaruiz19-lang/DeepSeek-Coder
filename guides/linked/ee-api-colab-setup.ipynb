```
#!/usr/bin/env python3
"""
Módulo de conexión y monitoreo para la CLI de Arduino.
Establece comunicación serial y gestiona los comandos de entrada/salida.
"""

import serial
import serial.tools.list_ports
import time
import sys

class ArduinoCLIBridge:
    def __init__(self, baudrate=115200, timeout=1):
        self.baudrate = baudrate
        self.timeout = timeout
        self.port = None
        self.connection = None

    def find_arduino(self):
        """Busca automáticamente un puerto serial conectado a Arduino."""
        ports = serial.tools.list_ports.comports()
        for port in ports:
            if "Arduino" in port.description or "USB-SERIAL" in port.description or "CH340" in port.description:
                print(f"[+] Dispositivo encontrado: {port.device} ({port.description})")
                return port.device
        if ports:
            # Si no encuentra coincidencia exacta, toma el primero disponible
            print(f"[!] Usando puerto por defecto: {ports[0].device}")
            return ports[0].device
        return None

    def connect(self, port=None):
        """Conecta al puerto serie especificado o autodetectado."""
        if not port:
            port = self.find_arduino()

        if not port:
            raise ConnectionError("No se encontró ningún puerto serie activo para Arduino.")

        try:
            self.connection = serial.Serial(port, self.baudrate, timeout=self.timeout)
            time.sleep(2) # Esperar reseteo por hardware de Arduino
            self.port = port
            print(f"[OK] Conectado exitosamente a {port} a {self.baudrate} baudios.")
        except Exception as e:
            print(f"[-] Error al conectar con el puerto {port}: {e}")
            sys.exit(1)

    def send_command(self, command: str):
        """Envía un comando a la CLI del Arduino."""
        if self.connection and self.connection.is_open:
            full_command = command.strip() + "\n"
            self.connection.write(full_command.encode('utf-8'))
            time.sleep(0.05)
            response = self.read_response()
            return response
        return "Error: Conexión no establecida."

    def read_response(self):
        """Lee la respuesta disponible desde el buffer serial."""
        output = []
        while self.connection.in_waiting > 0:
            line = self.connection.readline().decode('utf-8', errors='ignore').strip()
            if line:
                output.append(line)
        return "\n".join(output)

    def close(self):
        """Cierra la conexión serial."""
        if self.connection and self.connection.is_open:
            self.connection.close()
            print("[*] Conexión serial cerrada.")

if __name__ == "__main__":
    bridge = ArduinoCLIBridge()
    bridge.connect()
    print(bridge.send_command("STATUS"))
    bridge.close()
```

```
#!/usr/bin/env python3
"""
Simulador de Curvatura Espaciotemporal basada en la Velocidad Lumínica.
Calcula métricas de dilatación temporal y desfase para la transmisión de datos.
"""

import math
import time

class SpacetimeCurvatureEngine:
    def __init__(self, mass_kg=5.972e24):
        # Constantes físicas fundamentales
        self.C = 299792458.0  # Velocidad de la luz en el vacío (m/s)
        self.G = 6.67430e-11  # Constante gravitacional (m^3 kg^-1 s^-2)
        self.mass = mass_kg   # Masa de referencia (por defecto la Tierra)

    def calculate_schwarzschild_radius(self):
        """Calcula el radio de Schwarzschild para la masa dada."""
        return (2 * self.G * self.mass) / (self.C ** 2)

    def calculate_time_dilation_factor(self, distance_meters: float):
        """
        Calcula el factor de dilatación temporal (Gamma)
        en función de la distancia al centro de masa.
        """
        rs = self.calculate_schwarzschild_radius()
        if distance_meters <= rs:
            raise ValueError("La distancia está dentro del horizonte de sucesos.")

        # Métrica de Schwarzschild simplificada para dilatación temporal
        factor = math.sqrt(1.0 - (rs / distance_meters))
        return factor

    def modulate_signal_latency(self, base_delay: float, distance_meters: float):
        """
        Calcula la latencia ajustada aplicando la curvatura espaciotemporal
        sobre la velocidad lumínica.
        """
        dilation = self.calculate_time_dilation_factor(distance_meters)
        # El tiempo de propagación de la luz ajustado por la curvatura
        effective_c = self.C * dilation
        adjusted_latency = base_delay * (self.C / effective_c)
        return adjusted_latency

if __name__ == "__main__":
    engine = SpacetimeCurvatureEngine()
    rs = engine.calculate_schwarzschild_radius()
    print(f"[*] Radio de Schwarzschild calculado: {rs:.4f} metros")

    # Simular a una distancia orbital de 10,000,000 metros del centro de masa
    dist = 1.0e7
    factor = engine.calculate_time_dilation_factor(dist)
    print(f"[*] Factor de dilatación temporal a {dist}m: {factor:.8f}")
```

```
#!/usr/bin/env python3
"""
Módulo de Aceleración Relativista para Motor Brushless (Pin Echo).
Simula la manipulación de potencia y algoritmos de campo electromagnético
para alcanzar umbrales lumínicos virtuales.
"""

import math
import time

class BrushlessLightspeedEngine:
    def __init__(self, baseline_kv=2300, target_pin="ECHO"):
        self.c = 299792458.0  # Velocidad de la luz en m/s
        self.baseline_kv = baseline_kv  # RPM por voltio nominal del motor brushless
        self.target_pin = target_pin
        self.current_velocity = 0.0
        self.lorentz_factor = 1.0

    def calculate_relativistic_boost(self, throttle_percentage: float, iterations: int):
        """
        Aplica un algoritmo de transformación de Lorentz modificado para
        simular el incremento exponencial de potencia hacia el límite de c.
        """
        beta = throttle_percentage / 100.0
        if beta >= 1.0:
            beta = 0.999999  # Evitar singularidad matemática exacta en c

        # Factor de Lorentz: gamma = 1 / sqrt(1 - v^2/c^2)
        self.lorentz_factor = 1.0 / math.sqrt(1.0 - (beta ** 2))

        # Velocidad simulada del campo magnético del estator (fracción de c)
        simulated_speed = beta * self.c

        # Potencia efectiva multiplicada por el tensor de curvatura de iteración
        effective_power = self.baseline_kv * self.lorentz_factor * (1.0 + (iterations * 0.05))

        return simulated_speed, effective_power

    def generate_esc_pwm_signal(self, effective_power: float):
        """
         Traduce la potencia lumínica calculada a un ancho de pulso PWM
        compatible con el pin ECHO del microcontrolador.
        """
        # Mapeo escalar de seguridad para la interfaz de hardware
        pwm_microseconds = min(2000, max(1000, 1000 + (effective_power / 1000.0)))
        return int(pwm_microseconds)

if __name__ == "__main__":
    print("=======================================================")
    print(" INICIANDO PROPULSOR BRUSHLESS RELATIVISTA (PIN ECHO)")
    print("=======================================================")

    engine = BrushlessLightspeedEngine(baseline_kv=3000)

    # Simular rampa de aceleración hacia la velocidad lumínica
    for step in range(1, 6):
        throttle = step * 18.0  # Incrementando porcentaje de aceleración
        speed, power = engine.calculate_relativistic_boost(throttle, step)
        pwm = engine.generate_esc_pwm_signal(power)

        print(f"\n[Iteración {step}] Aceleración al {throttle:.1f}%:")
        print(f"  -> Pin destino: {engine.target_pin}")
        print(f"  -> Velocidad de campo simulada: {speed:,.2f} m/s")
        print(f"  -> Factor de Lorentz (γ): {engine.lorentz_factor:.4f}")
        print(f"  -> Pulso ESC calculado: {pwm} µs")
        time.sleep(0.3)

    print("\n[OK] Algoritmos físicos ajustados y sincronizados con éxito.")
```

```
#!/usr/bin/env python3
"""
Módulo de Modulación Vinyl Scratch para la placa LM-AK135L (Pin Echo).
Simula efectos de DJ (scratch, back-spin, crossfader dinámico)
sobre el flujo de control electromagnético.
"""

import math
import time

class VinylScratchModulator:
    def __init__(self, board_model="LM-AK135L", target_pin="ECHO"):
        self.board_model = board_model
        self.target_pin = target_pin
        self.scratch_intensity = 0.85
        self.turntable_rpm = 33.33

    def generate_scratch_envelope(self, time_step: float, scratch_type: "str" = "transformer"):
        """
        Calcula la modulación de frecuencia y amplitud basada en curvas
        de scratch de vinilo (movimiento sinusoidal amortiguado con inversión de fase).
        """
        if scratch_type == "baby_scratch":
            # Movimiento lineal de ida y vuelta
            envelope = math.sin(time_step * 10.0)
        elif scratch_type == "flare":
            # Cortes rápidos tipo crossfader
            base = math.sin(time_step * 15.0)
            envelope = base * (1.0 if abs(base) > 0.3 else -0.5)
        else: # "transformer" o por defecto (efecto complejo de alta velocidad)
            envelope = math.sin(time_step * 20.0) * math.cos(time_step * 5.0)

        return envelope

    def apply_scratch_to_pwm(self, base_pwm: int, time_step: float, scratch_type: "str" = "transformer"):
        """
        Aplica el efecto de scratch modificando el ancho de pulso PWM
        en el pin ECHO de la placa LM-AK135L.
        """
        envelope = self.generate_scratch_envelope(time_step, scratch_type)

        # Modulación de modulación: el scratch invierte temporalmente el sentido del pulso
        modulated_pwm = int(base_pwm + (envelope * 400.0 * self.scratch_intensity))

        # Asegurar límites seguros para la señal de control del hardware
        return max(500, min(2500, modulated_pwm)), envelope

if __name__ == "__main__":
    print(f"=======================================================")
    print(f" MODULADOR DJ VINYL SCRATCH - PLACA {VinylScratchModulator().board_model}")
    print(f"=======================================================")

    modulator = VinylScratchModulator()

    # Simular una sesión de scratch de 5 etapas en el pin ECHO
    base_signal_pwm = 1500  # Pulso neutral en microsegundos

    for i in range(1, 6):
        t = i * 0.2
        st_type = "flare" if i % 2 == 0 else "baby_scratch"
        new_pwm, env = modulator.apply_scratch_to_pwm(base_signal_pwm, t, st_type)

        print(f"\n[Scratch Step {i}] Tipo: {st_type}")
        print(f"  -> Pin destino: {modulator.target_pin} (Placa {modulator.board_model})")
        print(f"  -> Coeficiente de envolvente (Envolvente DJ): {env:+.4f}")
        print(f"  -> PWM Modulado resultante: {new_pwm} µs")
        time.sleep(0.2)

    print("\n[OK] Efecto scratch sincronizado correctamente con la arquitectura de la placa.")
```

```
#!/usr/init/env python3
"""
Módulo de escaneo y conexión Bluetooth para la placa LM-AK135L
filtrando exclusivamente por nombre de dispositivo.
"""

import serial.tools.list_ports
import time

class BluetoothNameConnector:
    def __init__(self, target_bluetooth_name="LM-AK135L-CORE"):
        self.target_name = target_bluetooth_name
        self.active_port = None

    def scan_and_connect(self):
        print(f"[*] Buscando dispositivos Bluetooth con el nombre: '{self.target_name}'...")

        while not self.active_port:
            ports = serial.tools.list_ports.comports()
            for port in ports:
                # Verificamos si la descripción o el hardware ID coinciden con el nombre Bluetooth
                if self.target_name.lower() in port.description.lower() or self.target_name.lower() in port.device.lower():
                    self.active_port = port.device
                    print(f"[OK] ¡Dispositivo encontrado!")
                    print(f"  -> Nombre/ID: {port.description}")
                    print(f"  -> Puerto asignado: {self.active_port}")
                    return self.active_port

            print("[-] Dispositivo no visible. Reintentando escaneo en 2 segundos...")
            time.sleep(2)

        return self.active_port

if __name__ == "__main__":
    connector = BluetoothNameConnector(target_bluetooth_name="LM-AK135L")
    port = connector.scan_and_connect()
    print(f"[*] Canal Bluetooth establecido de forma segura en: {port}")
```

```
#!/usr/bin/env python3
"""
======================================================================
MOTOR DE SIMULACIÓN Y RECUPERACIÓN DE MATRICES (RESTORE ENGINE)
======================================================================
Simula la fragmentación, procesamiento dinámico y reensamblaje
estructural de matrices de datos hacia su estado original base.
"""

import time
import hashlib
import random

class MatrixNode:
    """Representa un nodo individual dentro de una estructura matricial."""
    def __init__(self, node_id: int, payload: dict):
        self.node_id = node_id
        self.payload = payload
        self.integrity_checksum = self._generate_checksum(payload)

    def _generate_checksum(self, data: dict) -> str:
        """Genera un hash criptográfico único para verificar la integridad del nodo."""
        data_str = str(sorted(data.items()))
        return hashlib.sha256(data_str.encode('utf-8')).hexdigest()[:12]

class RestorationSimulator:
    """Gestor de desintegración virtual y reensamblaje de matrices."""
    def __init__(self, matrix_name: str):
        self.matrix_name = matrix_name
        self.original_state = []
        self.fragmented_state = []

    def load_baseline_state(self, nodes: list):
        """Registra el estado base original antes de cualquier alteración."""
        self.original_state = nodes
        print(f"[*] Matriz '{self.matrix_name}' cargada en el buffer con {len(nodes)} nodos base.")

    def simulate_fragmentation(self):
        """Simula la disociación y dispersión de los nodos de la matriz."""
        print(f"\n[!] Iniciando protocolo de disociación en la matriz {self.matrix_name}...")
        self.fragmented_state = list(self.original_state)
        random.shuffle(self.fragmented_state)
        for node in self.fragmented_state:
            print(f"    - Nodo {node.node_id} disociado (Checksum: {node.integrity_checksum})")
        time.sleep(0.5)

    def execute_restoration_and_reassembly(self):
        """Reconstruye y alinea los nodos fragmentados hacia el estado original exacto."""
        print(f"\n[*] Ejecutando escaneo y reensamblaje estructural de precisión...")

        reconstructed_nodes = []
        # Ordenar los nodos utilizando el checksum original y el ID de nodo
        for original in self.original_state:
            found = next((n for n in self.fragmented_state if n.node_id == original.node_id and n.integrity_checksum == original.integrity_checksum), None)
            if found:
                reconstructed_nodes.append(found)
                print(f"    [OK] Nodo {found.node_id} reinyectado y alineado con éxito.")
            time.sleep(0.1)

        self.fragmented_state = reconstructed_nodes
        print(f"\n[SUCCESS] Reensamblaje completado. La matriz '{self.matrix_name}' ha recuperado el 100% de su estructura original.")

if __name__ == "__main__":
    print("=======================================================")
    print(" SIMULADOR DE RESTAURACIÓN MATRICIAL - PROTOCOLO V2.6")
    print("=======================================================")

    # 1. Crear nodos con datos estructurados (simulando componentes de materia/datos)
    nodes_baseline = [
        MatrixNode(node_id=101, payload={"element": "Carbono", "valence": 4, "spin": 0.5}),
        MatrixNode(node_id=102, payload={"element": "Silicio", "valence": 4, "spin": 0.5}),
        MatrixNode(node_id=103, payload={"element": "Oxígeno", "valence": 2, "spin": 1.0}),
        MatrixNode(node_id=104, payload={"element": "Nitrógeno", "valence": 3, "spin": 1.5})
    ]

    # 2. Inicializar el motor de simulación
    simulator = RestorationSimulator(matrix_name="MATRIZ-CORE-ALPHA")

    # 3. Registrar estado inicial
    simulator.load_baseline_state(nodes_baseline)

    # 4. Simular disociación / fragmentación
    simulator.simulate_fragmentation()

    # 5. Ejecutar reensamblaje y retorno al estado original
    simulator.execute_restoration_and_reassembly()
```

```
#!/usr/bin/env python3
"""
Módulo de Simulación Supralumínica (v > c)
Permite escalar el motor virtual más allá de la velocidad de la luz.
"""

import math

class SuperluminalEngine:
    def __init__(self, base_c=299792458.0):
        self.c = base_c
        self.superluminal_multiplier = 1.05  # Factor inicial superior a 1 (ej: 1.05c)

    def engage_superluminal_drive(self, target_factor: float):
        """
        Escala la velocidad simulada por encima de la velocidad de la luz (c).
        """
        self.superluminal_multiplier = max(1.0, target_factor)
        effective_speed = self.c * self.superluminal_multiplier

        print(f"\n[⚡] ACTIVANDO IMPULSOR SUPRALUMÍNICO:")
        print(f"    - Constante Base (c): {self.c:,.2f} m/s")
        print(f"    - Factor de Escala Aplicado: {self.superluminal_multiplier}x")
        print(f"    - Velocidad Virtual Resultante: {effective_speed:,.2f} m/s")

        return effective_speed

if __name__ == "__main__":
    engine = SuperluminalEngine()
    # Simular un salto al doble de la velocidad de la luz en el entorno de código
    engine.engage_superluminal_drive(target_factor=2.0)
```

```
#!/usr/bin/env python3
"""
======================================================================
MOTOR DE CONVERGENCIA TOPOLÓGICA DIRIGIDA (LCTD)
======================================================================
Emula una ley física algorítmica donde los nodos moleculares
desacoplan su estructura local y convergen de forma ordenada
hacia un sitio de decisión propio especificado por el sistema.
"""

import time
import hashlib
import random

class AtomicNode:
    """Representa un nodo molecular con vector de posición y estado."""
    def __init__(self, node_id: int, element: str, valence: int):
        self.node_id = node_id
        self.element = element
        self.valence = valence
        self.current_location = "ORIGIN_ZONE"

    def update_location(self, new_location: str):
        self.current_location = new_location

class TopologicalLawEngine:
    """
    Implementa la Ley de Convergencia Topológica Dirigida (LCTD):
    Controla la separación molecular ordenada y el reensamblaje en destino.
    """
    def __init__(self, simulation_name: str):
        self.name = simulation_name
        self.nodes = []
        self.decision_target = "DESTINATION_SITE_ALPHA"

    def register_nodes(self, nodes: list):
        self.nodes = nodes
        print(f"[*] Sistema '{self.name}' inicializado con {len(nodes)} nodos moleculares.")

    def set_decision_target(self, target_node_name: str):
        """Define el sitio de decisión propia donde ocurrirá la unión."""
        self.decision_target = target_node_name
        print(f"[🎯] Sitio de decisión propio establecido en: [{self.decision_target}]")

    def execute_directed_law(self):
        """
        Ejecuta la LCTD:
        1. Desacoplamiento ordenado de la materia.
        2. Transmisión vectorial basada en la regla de convergencia.
        3. Reensamblaje atómico en el sitio de destino.
        """
        print(f"\n--- APLICANDO LEY DE CONVERGENCIA TOPOLÓGICA (LCTD) ---")

        # Fase 1: Separación molecular ordenada
        print("[1] Iniciando separación molecular controlada (desacoplamiento de enlaces)...")
        for node in self.nodes:
            node.update_location("VECTOR_DE_TRANSITO")
            print(f"    - Nodo {node.node_id} ({node.element}) separado y en tránsito.")
            time.sleep(0.1)

        # Fase 2: Cálculo de trayectoria según la ley algorítmica
        print("\n[2] Calculando vectores de reorientación hacia el sitio de decisión...")
        time.sleep(0.4)

        # Fase 3: Unión y reensamblaje atómico en el sitio elegido
        print(f"\n[3] Ejecutando convergencia atómica en '{self.decision_target}'...")
        for node in self.nodes:
            node.update_location(self.decision_target)
            print(f"    [OK] Nodo {node.node_id} ({node.element}) enlazado con éxito en {self.decision_target}.")
            time.sleep(0.1)

        print(f"\n[SUCCESS] Reensamblaje completado bajo la LCTD. Materia unificada en el nuevo sitio.")

if __name__ == "__main__":
    print("=======================================================")
    print(" SIMULADOR DE LEYES FÍSICAS EMERGENTES - LCTD")
    print("=======================================================")

    # 1. Definir nodos moleculares base
    molecular_structure = [
        AtomicNode(node_id=701, element="Carbono-Matriz", valence=4),
        AtomicNode(node_id=702, element="Silicio-Conductor", valence=4),
        AtomicNode(node_id=703, element="Oxígeno-Enlace", valence=2)
    ]

    # 2. Inicializar el motor de la ley topológica
    engine = TopologicalLawEngine(simulation_name="LCTD-CORE-01")
    engine.register_nodes(molecular_structure)

    # 3. Elegir el sitio de decisión propia para la unión
    engine.set_decision_target("SITIO_DESTINO_BETA_2")

    # 4. Ejecutar el ciclo de separación y reensamblaje dirigido
    engine.execute_directed_law()
```

The Earth Engine API is installed by default in Google Colaboratory so requires only importing and authenticating. These steps must be completed for each new Colab session, if you restart your Colab kernel, or if your Colab virtual machine is recycled due to inactivity.

```
import ee
```

## Test the API

Test the API by printing the elevation of Mount Everest.

```
# Print the elevation of Mount Everest.
dem = ee.Image('USGS/SRTMGL1_003')
xy = ee.Geometry.Point([86.9250, 27.9881])
elev = dem.sample(xy, 30).first().get('elevation').getInfo()
print('Mount Everest elevation (m):', elev)
```

## Map visualization

`ee.Image` objects can be displayed to notebook output cells. The following two
examples demonstrate displaying a static image and an interactive map.


### Static image

The `IPython.display` module contains the `Image` function, which can display
the results of a URL representing an image generated from a call to the Earth
Engine `getThumbUrl` function. The following cell will display a thumbnail
of the global elevation model.

```
# Import the Image function from the IPython.display module.
from IPython.display import Image

# Display a thumbnail of global elevation.
Image(url = dem.updateMask(dem.gt(0))
  .getThumbURL({'min': 0, 'max': 4000, 'dimensions': 512,
                'palette': ['006633', 'E5FFCC', '662A00', 'D8D8D8', 'F5F5F5']}))
```

### Interactive map

The [geemap](https://github.com/gee-community/geemap)
library can be used to display `ee.Image` objects on an interactive
[ipyleaflet](https://github.com/jupyter-widgets/ipyleaflet) map.

The following cell provides an example of using the `geemap.Map` object to
display an elevation model.

```
# Import the geemap library.
import geemap

# Set visualization parameters.
vis_params = {
  'min': 0,
  'max': 4000,
  'palette': ['006633', 'E5FFCC', '662A00', 'D8D8D8', 'F5F5F5']}

# Create a map object.
m = geemap.Map(center=[20, 0], zoom=3)

# Add the elevation model to the map object.
m.add_ee_layer(dem.updateMask(dem.gt(0)), vis_params, 'DEM')

# Display the map.
display(m)
```

## Chart visualization

Some Earth Engine functions produce tabular data that can be plotted by
data visualization packages such as `matplotlib`. The following example
demonstrates the display of tabular data from Earth Engine as a scatter
plot. See [Charting in Colaboratory](https://colab.sandbox.google.com/notebooks/charts.ipynb)
for more information.

```
# Import the matplotlib.pyplot module.
import matplotlib.pyplot as plt

# Fetch a Landsat TOA image.
img = ee.Image('LANDSAT/LT05/C02/T1_TOA/LT05_034033_20000913')

# Select Red and NIR bands and sample 500 points.
samp_fc = img.select(['B3','B4']).sample(scale=30, numPixels=500)

# Arrange the sample as a list of lists.
samp_dict = samp_fc.reduceColumns(ee.Reducer.toList().repeat(2), ['B3', 'B4'])
samp_list = ee.List(samp_dict.get('list'))

# Save server-side ee.List as a client-side Python list.
samp_data = samp_list.getInfo()

# Display a scatter plot of Red-NIR sample pairs using matplotlib.
plt.scatter(samp_data[0], samp_data[1], alpha=0.2)
plt.xlabel('Red', fontsize=12)
plt.ylabel('NIR', fontsize=12)
plt.show()
```
